```python
import numpy as np
```

```python
def weighted_mark(marks, weights=np.array([0.3, 0.4, 0.2, 0.1])):
    return int(np.round(np.dot(marks, weights)))

assert weighted_mark(np.array([7, 10, 8, 6])) == 8
print(weighted_mark(np.array([7, 10, 8, 6])))
```

```python
def replace_every_third(a, value):
    res = a.copy()
    res[::3] = value
    return res

a = np.array([3, 5, 1, 0, -3, 22, 213436])
res = replace_every_third(a, -111)
assert np.array_equal(res, np.array([-111, 5, 1, -111, -3, 22, -111]))
print(res)
```

```python
def close_indices(a, b, eps):
    return np.flatnonzero(np.abs(a - b) <= eps)

a = np.array([1.5, 0.5, 2, -4.1, -3, 6, -1])
b = np.array([1.2, 0.5, 1, -4, 3, 0, -1.2])
res = close_indices(a, b, 0.5)
assert np.array_equal(res, np.array([0, 1, 3, 6])) and isinstance(res, np.ndarray) and res.ndim == 1
print(res)
```

```python
def block_matrix(m):
    return np.tile(m, (2, 2))

m = np.array([[0, 1, 2], [3, 4, 5]])
res = block_matrix(m)
assert np.array_equal(res, np.array([[0, 1, 2, 0, 1, 2],
                                     [3, 4, 5, 3, 4, 5],
                                     [0, 1, 2, 0, 1, 2],
                                     [3, 4, 5, 3, 4, 5]]))
print(res)
```

```python
def diag_nonzero_prod(m):
    d = np.diag(m)
    return int(np.prod(d[d != 0]))

m = np.arange(9).reshape(3, 3)
assert diag_nonzero_prod(m) == 32
print(diag_nonzero_prod(m))
```

```python
def normalize(m):
    m = m.astype(float)
    std = m.std(axis=0)
    std = np.where(std == 0, 1, std)
    return (m - m.mean(axis=0)) / std

m = np.array([[1, 4, 4200], [0, 10, 5000], [1, 2, 1000]])
res = normalize(m)
expected = np.array([[0.71, -0.39, 0.46], [-1.41, 1.37, 0.93], [0.71, -0.98, -1.39]])
assert np.allclose(np.round(res, 2), expected)

res_const = normalize(np.array([[1, 5], [2, 5], [3, 5]]))
assert not np.isnan(res_const).any()
print(np.round(res, 2))
print(res_const)
```

```python
def antiderivative(coefs, c=0):
    n = len(coefs)
    powers = np.arange(n, 0, -1)         
    return np.append(coefs / powers, c)

p = np.array([4, 6, 0, 1])
assert np.allclose(antiderivative(p), [1, 2, 0, 1, 0])
assert np.allclose(antiderivative(p, c=-2), [1, 2, 0, 1, -2])
print(antiderivative(p), antiderivative(p, c=-2))
```

```python
def make_symmetric(m):
    return m + m.T - np.diag(np.diag(m))

m = np.array([[1, 2, 3, 4], [0, 5, 6, 7], [0, 0, 8, 9], [0, 0, 0, 10]])
res = make_symmetric(m)
assert np.array_equal(res, np.array([[1, 2, 3, 4], [2, 5, 6, 7], [3, 6, 8, 9], [4, 7, 9, 10]]))
assert np.array_equal(res, res.T)
print(res)
```

```python
def range_matrix(m, a, b):
    return np.tile(np.arange(a, b + 1), (m, 1))

res = range_matrix(5, 3, 10)
assert res.shape == (5, 8) and np.array_equal(res[0], np.arange(3, 11)) and (res == res[0]).all()
print(res)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

a = np.array([-2, 1, 0, -5, 4, 3, -3])
b = np.array([0, 2, -2, 10, 6, 0, 0])
assert np.isclose(cosine_similarity(a, b), -0.25)
print(cosine_similarity(a, b))
```
